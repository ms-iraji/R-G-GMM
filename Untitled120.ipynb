```python
# ============================================================
# Residual-Guided GMM Undersampling for Gradient Boosting
# ============================================================
#
# FULL EXPERIMENTAL FRAMEWORK
#
# Methods:
#
# 1. Original Gradient Boosting
# 2. Random Undersampling + Gradient Boosting
# 3. Cluster Undersampling + Gradient Boosting
# 4. SMOTE + Gradient Boosting
# 5. Residual-Guided GMM Undersampling for Gradient Boosting
#
# Validation:
#     Stratified 5-Fold × 5 Repeats
#
# IMPORTANT:
#     Sampling is performed ONLY on the training fold.
#
# Reporting:
#
# LEVEL 1:
#     Dataset-level
#
#     For every dataset:
#         Fold mean -> Repeat mean
#         Repeat means -> Dataset Mean ± SD
#
# LEVEL 2:
#     Overall across datasets
#
#     Average of Dataset-level means.
#
# Proposed wins are counted separately for:
#
#     Balanced Accuracy
#     G-Mean
#     F1
#     MCC
#     ROC-AUC
#
# ============================================================


# ============================================================
# 0. INSTALL REQUIRED PACKAGE
# ============================================================

!pip -q install imbalanced-learn


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import re
import math
import random
import shutil
import zipfile
import warnings
import urllib3

import numpy as np
import pandas as pd

from collections import Counter

from scipy.special import expit as sigmoid

from sklearn.tree import DecisionTreeRegressor
from sklearn.mixture import GaussianMixture
from sklearn.cluster import KMeans
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    roc_auc_score,
    confusion_matrix
)

from imblearn.over_sampling import SMOTE

import requests
from bs4 import BeautifulSoup


warnings.filterwarnings("ignore")

urllib3.disable_warnings(
    urllib3.exceptions.InsecureRequestWarning
)


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

BASE_SEED = 42

N_REPEATS = 3
N_FOLDS = 2

N_ITER = 100

LEARNING_RATE = 0.05

TREE_DEPTH = 4

THRESHOLD = 0.1

BALANCE_RATE = 0.5


np.random.seed(BASE_SEED)
random.seed(BASE_SEED)


# ============================================================
# 3. ALL DATASETS
# ============================================================

ALL_DATASETS = [

    "ecoli1",

    "glass1",

    "yeast1",

    "vehicle0",

    "page-blocks0",

    "pima",

    "abalone19",

    "winequality-red-4",

    "shuttle-2_vs_5"
]


# ============================================================
# 4. ASK USER HOW MANY DATASETS TO RUN
# ============================================================

print()
print("=" * 90)
print("RESIDUAL-GUIDED GMM EXPERIMENT")
print("=" * 90)

print()
print(
    f"Total number of datasets available for the experiment: "
    f"{len(ALL_DATASETS)}"
)

print()
print("Available datasets:")

for i, name in enumerate(
    ALL_DATASETS,
    start=1
):

    print(
        f"{i}. {name}"
    )


print()
print(
    f"Validation: Stratified {N_FOLDS}-Fold × "
    f"{N_REPEATS} Repeats"
)

print(
    f"Total folds per dataset: "
    f"{N_FOLDS * N_REPEATS}"
)

print()


while True:

    try:

        N_DATASETS_TO_RUN = int(
            input(
                f"How many datasets do you want to run? "
                f"(1-{len(ALL_DATASETS)}): "
            )
        )

        if (
            1
            <=
            N_DATASETS_TO_RUN
            <=
            len(ALL_DATASETS)
        ):

            break

        print(
            f"Please enter a number between "
            f"1 and {len(ALL_DATASETS)}."
        )

    except ValueError:

        print(
            "Please enter a valid integer."
        )


# ------------------------------------------------------------
# Select WHICH N datasets to run
# ------------------------------------------------------------

while True:

    try:

        selected_input = input(
            f"Enter the dataset number(s) to run "
            f"(exactly {N_DATASETS_TO_RUN}, comma-separated; "
            f"e.g. 1,3,5): "
        ).strip()

        selected_numbers = [
            int(x.strip())
            for x in selected_input.split(",")
            if x.strip()
        ]

        if len(selected_numbers) != N_DATASETS_TO_RUN:
            print(
                f"Please select exactly "
                f"{N_DATASETS_TO_RUN} dataset(s)."
            )
            continue

        if any(
            number < 1 or number > len(ALL_DATASETS)
            for number in selected_numbers
        ):
            print(
                f"Dataset numbers must be between "
                f"1 and {len(ALL_DATASETS)}."
            )
            continue

        if len(set(selected_numbers)) != len(selected_numbers):
            print(
                "Duplicate dataset numbers are not allowed."
            )
            continue

        DATASETS = [
            ALL_DATASETS[number - 1]
            for number in selected_numbers
        ]

        break

    except ValueError:

        print(
            "Invalid input. Please enter dataset numbers "
            "separated by commas, e.g. 1,3,5."
        )


print()
print("=" * 90)

print(
    f"Selected {len(DATASETS)} dataset(s):"
)

for number, name in zip(
    selected_numbers,
    DATASETS
):

    print(
        f"{number}. {name}"
    )

print("=" * 90)


# ============================================================
# 5. KEEL WEBSITE
# ============================================================

KEEL_PAGE = (
    "https://sci2s.ugr.es/keel/imbalanced.php"
)

DATA_ROOT = "./keel_datasets"

os.makedirs(
    DATA_ROOT,
    exist_ok=True
)


# ============================================================
# 6. DOWNLOAD KEEL PAGE
# ============================================================

def download_keel_page():

    print()
    print("=" * 90)
    print("Downloading KEEL dataset page")
    print("=" * 90)

    response = requests.get(
        KEEL_PAGE,
        verify=False,
        timeout=60
    )

    response.raise_for_status()

    print(
        "KEEL page downloaded successfully."
    )

    return response.text


# ============================================================
# 7. FIND DATASET ZIP LINK
# ============================================================

def find_dataset_link(
    html,
    dataset_name
):

    soup = BeautifulSoup(
        html,
        "html.parser"
    )

    links = soup.find_all(
        "a",
        href=True
    )

    candidates = []

    dataset_normalized = (
        dataset_name
        .lower()
        .replace("_", "-")
    )

    for link in links:

        href = link.get(
            "href",
            ""
        )

        text = link.get_text(
            " ",
            strip=True
        )

        combined = (
            href
            + " "
            + text
        ).lower()

        combined = combined.replace(
            "_",
            "-"
        )

        if (
            dataset_normalized in combined
            and
            (
                ".zip" in combined
                or
                "data" in combined
            )
        ):

            candidates.append(
                href
            )


    if not candidates:

        return None


    for candidate in candidates:

        if ".zip" in candidate.lower():

            return candidate


    return candidates[0]


# ============================================================
# 8. ABSOLUTE URL
# ============================================================

def absolute_url(
    href
):

    if href.startswith(
        "http://"
    ) or href.startswith(
        "https://"
    ):

        return href

    if href.startswith("/"):

        return (
            "https://sci2s.ugr.es"
            + href
        )

    return (
        "https://sci2s.ugr.es/keel/"
        + href
    )


# ============================================================
# 9. DOWNLOAD AND EXTRACT DATASET
# ============================================================

def download_dataset(
    dataset_name,
    html
):

    dataset_dir = os.path.join(
        DATA_ROOT,
        dataset_name
    )

    os.makedirs(
        dataset_dir,
        exist_ok=True
    )


    # --------------------------------------------------------
    # Check existing .dat
    # --------------------------------------------------------

    existing_dat = []

    for root, dirs, files in os.walk(
        dataset_dir
    ):

        for file in files:

            if file.lower().endswith(
                ".dat"
            ):

                existing_dat.append(
                    os.path.join(
                        root,
                        file
                    )
                )


    if existing_dat:

        print(
            f"{dataset_name}: already available."
        )

        return existing_dat[0]


    # --------------------------------------------------------
    # Find ZIP
    # --------------------------------------------------------

    href = find_dataset_link(
        html,
        dataset_name
    )


    if href is None:

        raise FileNotFoundError(
            f"Could not find ZIP link for "
            f"{dataset_name} on KEEL page."
        )


    url = absolute_url(
        href
    )


    print()
    print(
        f"Downloading {dataset_name}"
    )

    print(
        f"URL: {url}"
    )


    zip_path = os.path.join(
        dataset_dir,
        dataset_name + ".zip"
    )


    # --------------------------------------------------------
    # Download
    # --------------------------------------------------------

    response = requests.get(
        url,
        verify=False,
        timeout=120
    )

    response.raise_for_status()


    with open(
        zip_path,
        "wb"
    ) as f:

        f.write(
            response.content
        )


    # --------------------------------------------------------
    # Extract
    # --------------------------------------------------------

    with zipfile.ZipFile(
        zip_path,
        "r"
    ) as z:

        z.extractall(
            dataset_dir
        )


    # --------------------------------------------------------
    # Locate .dat
    # --------------------------------------------------------

    dat_files = []

    for root, dirs, files in os.walk(
        dataset_dir
    ):

        for file in files:

            if file.lower().endswith(
                ".dat"
            ):

                dat_files.append(
                    os.path.join(
                        root,
                        file
                    )
                )


    if not dat_files:

        raise FileNotFoundError(
            f"No .dat file found for "
            f"{dataset_name}"
        )


    print(
        f"{dataset_name}: extracted successfully."
    )


    return dat_files[0]


# ============================================================
# 10. DOWNLOAD SELECTED DATASETS
# ============================================================

html = download_keel_page()

DATASET_PATHS = {}

DOWNLOAD_ERRORS = {}


for dataset_name in DATASETS:

    try:

        DATASET_PATHS[
            dataset_name
        ] = download_dataset(
            dataset_name,
            html
        )

    except Exception as e:

        DOWNLOAD_ERRORS[
            dataset_name
        ] = str(e)

        print()
        print(
            f"ERROR downloading "
            f"{dataset_name}:"
        )

        print(e)


print()
print("=" * 90)
print("DATASET PATHS")
print("=" * 90)

for name in DATASETS:

    if name in DATASET_PATHS:

        print(
            f"{name:<25} -> "
            f"{DATASET_PATHS[name]}"
        )

    else:

        print(
            f"{name:<25} -> NOT AVAILABLE"
        )


# ============================================================
# 11. KEEL DATASET LOADER
# ============================================================

def load_keel_dataset(
    path
):

    attributes = []

    attribute_types = []

    data = []

    reading_data = False


    with open(
        path,
        "r",
        encoding="utf-8",
        errors="ignore"
    ) as file:

        for line in file:

            line = line.strip()

            if not line:

                continue


            lower = line.lower()


            # ------------------------------------------------
            # Attribute
            # ------------------------------------------------

            if lower.startswith(
                "@attribute"
            ):

                parts = line.split(
                    maxsplit=2
                )


                if len(parts) >= 3:

                    attr_name = (
                        parts[1]
                        .strip()
                    )

                    attr_type = (
                        parts[2]
                        .strip()
                    )


                    attributes.append(
                        attr_name
                    )

                    attribute_types.append(
                        attr_type
                    )


                continue


            # ------------------------------------------------
            # Data
            # ------------------------------------------------

            if lower == "@data":

                reading_data = True

                continue


            # ------------------------------------------------
            # Rows
            # ------------------------------------------------

            if reading_data:

                if line.startswith(
                    "%"
                ):

                    continue


                row = [
                    item.strip()
                    for item in line.split(",")
                ]


                data.append(
                    row
                )


    dataframe = pd.DataFrame(
        data,
        columns=attributes
    )


    # --------------------------------------------------------
    # Strip whitespace
    # --------------------------------------------------------

    for column in dataframe.columns:

        if dataframe[
            column
        ].dtype == object:

            dataframe[
                column
            ] = (
                dataframe[
                    column
                ]
                .astype(str)
                .str.strip()
            )


    return dataframe


# ============================================================
# 12. PREPARE DATASET
# ============================================================

def prepare_dataset(
    path
):

    df = load_keel_dataset(
        path
    )


    # --------------------------------------------------------
    # Target column
    # --------------------------------------------------------

    target_candidates = [
        c
        for c in df.columns
        if c.lower()
        in [
            "class",
            "target",
            "label"
        ]
    ]


    if target_candidates:

        target_col = (
            target_candidates[-1]
        )

    else:

        target_col = (
            df.columns[-1]
        )


    # --------------------------------------------------------
    # Target
    # --------------------------------------------------------

    target = (
        df[target_col]
        .astype(str)
        .str.strip()
    )


    # --------------------------------------------------------
    # Remove missing target
    # --------------------------------------------------------

    valid_target = (
        target.notna()
        &
        (
            target.str.lower()
            != "?"
        )
    )


    df = df.loc[
        valid_target
    ].copy()


    target = (
        df[target_col]
        .astype(str)
        .str.strip()
    )


    # --------------------------------------------------------
    # Classes
    # --------------------------------------------------------

    classes = list(
        pd.unique(
            target
        )
    )


    if len(classes) != 2:

        raise ValueError(
            f"Dataset must be binary. "
            f"Found classes: {classes}"
        )


    class_counts = (
        target.value_counts()
    )


    minority_original = (
        class_counts.idxmin()
    )


    majority_original = (
        class_counts.idxmax()
    )


    # --------------------------------------------------------
    # Minority = 1
    # Majority = 0
    # --------------------------------------------------------

    y = np.where(
        target.values
        ==
        minority_original,
        1,
        0
    ).astype(int)


    # --------------------------------------------------------
    # Features
    # --------------------------------------------------------

    X_df = df.drop(
        columns=[
            target_col
        ]
    ).copy()


    # --------------------------------------------------------
    # Convert features
    # --------------------------------------------------------

    for column in X_df.columns:

        numeric_version = pd.to_numeric(
            X_df[column],
            errors="coerce"
        )


        numeric_ratio = (
            numeric_version.notna()
            .mean()
        )


        if numeric_ratio >= 0.95:

            X_df[column] = (
                numeric_version
            )

        else:

            X_df[column] = (
                X_df[column]
                .astype(str)
                .str.strip()
                .astype("category")
                .cat.codes
            )


    # --------------------------------------------------------
    # Replace infinity
    # --------------------------------------------------------

    X_df = X_df.replace(
        [
            np.inf,
            -np.inf
        ],
        np.nan
    )


    # --------------------------------------------------------
    # Median imputation
    # --------------------------------------------------------

    for column in X_df.columns:

        if X_df[
            column
        ].isna().any():

            median_value = (
                X_df[
                    column
                ].median()
            )


            if pd.isna(
                median_value
            ):

                median_value = 0.0


            X_df[
                column
            ] = X_df[
                column
            ].fillna(
                median_value
            )


    X = X_df.astype(
        float
    ).values


    # --------------------------------------------------------
    # Information
    # --------------------------------------------------------

    counter = Counter(
        y
    )


    print()
    print("=" * 90)
    print("DATASET INFORMATION")
    print("=" * 90)

    print(
        "Dataset:",
        os.path.basename(path)
    )

    print(
        "Samples:",
        X.shape[0]
    )

    print(
        "Features:",
        X.shape[1]
    )

    print(
        "Original classes:",
        classes
    )

    print(
        "Original class distribution:",
        dict(class_counts)
    )

    print(
        "Encoded distribution:",
        dict(counter)
    )

    print(
        "Minority class:",
        minority_original,
        "-> 1"
    )

    print(
        "Majority class:",
        majority_original,
        "-> 0"
    )

    print(
        "Imbalance ratio:",
        round(
            counter[0] /
            counter[1],
            4
        )
    )


    return (
        X,
        y,
        1,
        0
    )


# ============================================================
# 13. METRICS
# ============================================================

def calculate_metrics(
    y_true,
    y_pred,
    y_prob,
    minority_class
):

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[
            0,
            1
        ]
    )


    TN, FP, FN, TP = (
        cm.ravel()
    )


    sensitivity = (
        TP / (TP + FN)
        if (TP + FN) > 0
        else 0.0
    )


    specificity = (
        TN / (TN + FP)
        if (TN + FP) > 0
        else 0.0
    )


    accuracy = accuracy_score(
        y_true,
        y_pred
    )


    balanced_accuracy = (
        sensitivity
        +
        specificity
    ) / 2.0


    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )


    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )


    gmean_value = math.sqrt(
        max(
            0.0,
            sensitivity
            *
            specificity
        )
    )


    mcc = matthews_corrcoef(
        y_true,
        y_pred
    )


    try:

        auc_value = roc_auc_score(
            y_true,
            y_prob
        )

    except ValueError:

        auc_value = np.nan


    if minority_class == 1:

        minority_recall = (
            sensitivity
        )

    else:

        minority_recall = (
            specificity
        )


    return {

        "Accuracy":
            accuracy,

        "Balanced_Accuracy":
            balanced_accuracy,

        "GMean":
            gmean_value,

        "F1":
            f1,

        "MCC":
            mcc,

        "ROC_AUC":
            auc_value,

        "Sensitivity":
            sensitivity,

        "Specificity":
            specificity,

        "Minority_Recall":
            minority_recall,

        "TN":
            TN,

        "FP":
            FP,

        "FN":
            FN,

        "TP":
            TP
    }


# ============================================================
# 14. STANDARD GRADIENT BOOSTING
# ============================================================

def train_standard_gb(
    X_train,
    y_train,
    X_test,
    seed
):

    model = GradientBoostingClassifier(
        n_estimators=N_ITER,
        learning_rate=LEARNING_RATE,
        max_depth=TREE_DEPTH,
        random_state=seed
    )


    model.fit(
        X_train,
        y_train
    )


    pred = model.predict(
        X_test
    )


    prob = model.predict_proba(
        X_test
    )[:, 1]


    return (
        pred,
        prob
    )


# ============================================================
# 15. RANDOM UNDERSAMPLING
# ============================================================

def random_undersample(
    X,
    y,
    seed
):

    rng = np.random.RandomState(
        seed
    )


    indices_0 = np.where(
        y == 0
    )[0]


    indices_1 = np.where(
        y == 1
    )[0]


    n_minority = min(
        len(indices_0),
        len(indices_1)
    )


    selected_0 = rng.choice(
        indices_0,
        size=n_minority,
        replace=False
    )


    selected_1 = rng.choice(
        indices_1,
        size=n_minority,
        replace=False
    )


    selected = np.concatenate(
        [
            selected_0,
            selected_1
        ]
    )


    rng.shuffle(
        selected
    )


    return (
        X[selected],
        y[selected]
    )


# ============================================================
# 16. CLUSTER UNDERSAMPLING
# ============================================================

def cluster_undersample(
    X,
    y,
    seed
):

    rng = np.random.RandomState(
        seed
    )


    majority_idx = np.where(
        y == 0
    )[0]


    minority_idx = np.where(
        y == 1
    )[0]


    n_minority = len(
        minority_idx
    )


    n_majority = len(
        majority_idx
    )


    if n_majority <= n_minority:

        return (
            X.copy(),
            y.copy()
        )


    n_clusters = min(
        n_minority,
        n_majority
    )


    majority_X = X[
        majority_idx
    ]


    kmeans = KMeans(
        n_clusters=n_clusters,
        random_state=seed,
        n_init=10
    )


    labels = kmeans.fit_predict(
        majority_X
    )


    selected_majority = []


    for cluster_id in range(
        n_clusters
    ):

        cluster_positions = np.where(
            labels == cluster_id
        )[0]


        if len(
            cluster_positions
        ) == 0:

            continue


        cluster_points = (
            majority_X[
                cluster_positions
            ]
        )


        center = (
            kmeans.cluster_centers_[
                cluster_id
            ]
        )


        distances = np.linalg.norm(
            cluster_points
            -
            center,
            axis=1
        )


        best_position = (
            cluster_positions[
                np.argmin(
                    distances
                )
            ]
        )


        selected_majority.append(
            majority_idx[
                best_position
            ]
        )


    # --------------------------------------------------------
    # Safety
    # --------------------------------------------------------

    if len(
        selected_majority
    ) < n_minority:

        remaining = list(
            set(
                majority_idx
            )
            -
            set(
                selected_majority
            )
        )


        extra = rng.choice(
            remaining,
            size=(
                n_minority
                -
                len(
                    selected_majority
                )
            ),
            replace=False
        )


        selected_majority.extend(
            extra.tolist()
        )


    selected_majority = np.asarray(
        selected_majority
    )[:n_minority]


    selected = np.concatenate(
        [
            selected_majority,
            minority_idx
        ]
    )


    rng.shuffle(
        selected
    )


    return (
        X[selected],
        y[selected]
    )


# ============================================================
# 17. SMOTE
# ============================================================

def smote_resample(
    X,
    y,
    seed
):

    minority_count = np.sum(
        y == 1
    )


    if minority_count < 2:

        return (
            X.copy(),
            y.copy()
        )


    k_neighbors = min(
        5,
        minority_count - 1
    )


    smote = SMOTE(
        random_state=seed,
        k_neighbors=k_neighbors
    )


    X_resampled, y_resampled = (
        smote.fit_resample(
            X,
            y
        )
    )


    return (
        X_resampled,
        y_resampled
    )


# ============================================================
# 18. PROPOSED METHOD
#
# Residual-Guided GMM Undersampling
# for Gradient Boosting
# Tail-Capped Residual Selection (new variant)
#
# CORE LOGIC:
#
# GMM is applied to majority residual/error values.
#
# From each GMM component, samples are ranked by their
# distance from the component mean. The most extreme
# residual tail is excluded from the primary selection,
# and the second-tier residual band is preferred.
#
# Then:
#
#     selected majority
#     +
#     all minority
#
# are used for boosting.
# ============================================================

def train_proposed_cell60(
    X_train,
    y_train,
    X_test,
    repeat_seed,
    fold_idx
):

    instances = X_train

    labels = y_train


    base_classifier = DecisionTreeRegressor(
        max_depth=TREE_DEPTH
    )


    n_classifier = 100

    learning_rate = (
        LEARNING_RATE
    )

    learning_rate2 = 0.1

    N = 100

    rate = BALANCE_RATE


    XX = instances

    YY = labels


    X = list(XX)

    Y = list(YY)


    rng = np.random.RandomState(
        repeat_seed
        +
        fold_idx
    )


    Xg = XX

    yg = YY


    n_samplesg = Xg.shape[0]


    # --------------------------------------------------------
    # Initial prediction
    # --------------------------------------------------------

    y_pred_traing = rng.normal(
        size=n_samplesg
    )


    mean_y = np.mean(
        y
    )


    mean_y = np.clip(
        mean_y,
        1e-12,
        1 - 1e-12
    )


    ve = (
        0.5
        *
        np.log(
            (
                1
                +
                mean_y
            )
            /
            (
                1
                -
                mean_y
            )
        )
    )


    y_pred_traing = np.full(
        (
            n_samplesg,
        ),
        ve,
        dtype=float
    )


    # --------------------------------------------------------
    # Error initialization
    # --------------------------------------------------------

    errc = np.full(
        (
            n_samplesg,
        ),
        1.0 / len(X),
        dtype=float
    )


    predictors = []


    ytt = np.zeros(
        len(
            y_pred_traing
        )
    )


    # --------------------------------------------------------
    # Weight initialization
    # --------------------------------------------------------

    weight = []


    init_w = (
        1.0
        /
        len(X)
    )


    for i in range(
        len(X)
    ):

        weight.append(
            init_w
        )


    # ========================================================
    # BOOSTING LOOP
    #
    # PRESERVED FROM CELL 60
    # ========================================================

    h = 0


    while h <= N_ITER:


        # ----------------------------------------------------
        # Negative gradient
        # ----------------------------------------------------

        negative_gradient = -(
            sigmoid(
                y_pred_traing
            )
            -
            yg
        )


        new_predictor = (
            DecisionTreeRegressor(
                max_depth=TREE_DEPTH
            )
        )


        ytt = (
            negative_gradient
        )


        # ----------------------------------------------------
        # Determine majority
        # ----------------------------------------------------

        diff = (
            Y.count(1)
            >
            Y.count(0)
        )


        delete_list = []

        delete_list3 = []

        delete_list4 = []

        delete_listf = []

        keep_list = []


        # ----------------------------------------------------
        # Majority -> delete_list
        # Minority -> keep_list
        # ----------------------------------------------------

        if diff:

            for i in range(
                len(Y)
            ):

                if Y[i] == 1:

                    delete_data = [
                        i,
                        X[i],
                        1,
                        ytt[i],
                        errc[i]
                    ]

                    delete_list.append(
                        delete_data
                    )

                else:

                    keep_data = [
                        i,
                        X[i],
                        0,
                        ytt[i],
                        errc[i]
                    ]

                    keep_list.append(
                        keep_data
                    )

        else:

            for i in range(
                len(Y)
            ):

                if Y[i] == 0:

                    delete_data = [
                        i,
                        X[i],
                        0,
                        ytt[i],
                        errc[i]
                    ]

                    delete_list.append(
                        delete_data
                    )

                else:

                    keep_data = [
                        i,
                        X[i],
                        1,
                        ytt[i],
                        errc[i]
                    ]

                    keep_list.append(
                        keep_data
                    )


        # ----------------------------------------------------
        # Sort majority by error
        # ----------------------------------------------------

        delete_lists = sorted(
            delete_list,
            key=lambda x: x[4]
        )


        # ----------------------------------------------------
        # Residual values
        # ----------------------------------------------------

        Xk = [
            s[4]
            for s in delete_lists
        ]


        Xk = np.array(
            Xk,
            dtype=float
        )


        # ----------------------------------------------------
        # Feature vectors for the same majority samples, in the
        # SAME order as Xk / delete_lists. Standardized so that
        # Euclidean distance across the k-NN density estimate is
        # not dominated by any single high-variance feature.
        #
        # This is used ONLY for the local-density term below
        # (density measured in feature space = "how surrounded
        # this majority sample is by other majority samples",
        # rather than density of 1-D residual values, which is
        # largely redundant with the residual ranking itself).
        # ----------------------------------------------------

        Xk_feat_raw = np.array(
            [
                s[1]
                for s in delete_lists
            ],
            dtype=float
        )

        if Xk_feat_raw.shape[0] > 1:
            Xk_feat_std = StandardScaler().fit_transform(
                Xk_feat_raw
            )
        else:
            Xk_feat_std = Xk_feat_raw


        # ----------------------------------------------------
        # ----------------------------------------------------
        # ROBUST SECOND-TIER RESIDUAL-GUIDED GMM
        #
        # The original method selected the largest within-GMM
        # residual distances first. This can over-select extreme
        # outliers or samples close to class-overlap regions.
        #
        # New strategy:
        #   1. Select GMM components adaptively using BIC.
        #   2. Allocate the exact target_n majority samples
        #      proportionally across active components.
        #   3. Inside each component, rank samples by the
        #      standardized distance from the GMM component mean.
        #   4. Do NOT immediately select the most extreme samples.
        #      The upper extreme tail is temporarily excluded.
        #   5. Select primarily from the SECOND-TIER residual band:
        #         high residual, but not the most extreme residuals.
        #   6. Only use excluded extreme samples when a component
        #      does not contain enough non-extreme candidates.
        #   7. The final fallback follows the same robust
#      residual-density principle.
        #
        # Thus GMM still drives the selection, but the method
        # prefers informative high-residual samples rather than
        # blindly choosing the farthest/outlier samples.
        # ----------------------------------------------------

        target_n = len(keep_list)
        majority_n = len(delete_list)
        delete_listf = []

        # Fraction of the most extreme residuals temporarily
        # excluded from primary selection.
        EXTREME_TAIL_FRACTION = 0.10

        # Robust residual-density score:
        # higher residual is preferred, but very low-density
        # / isolated samples are penalized.
        RESIDUAL_SCORE_WEIGHT = 0.90
        DENSITY_SCORE_WEIGHT = 0.10

        # NEW:
        # Extreme residuals are soft-capped at the 90th
        # percentile instead of being hard-excluded.
        TAIL_CAP_QUANTILE = (
            1.0 - EXTREME_TAIL_FRACTION
        )

        if (
            target_n > 0
            and majority_n >= target_n
            and len(Xk) > 0
        ):

            max_components = min(
                10,
                target_n,
                majority_n
            )

            if max_components >= 4:
                min_components = 3
            else:
                min_components = 2

            candidate_components = range(
                min_components,
                max_components + 1
            )

            # ------------------------------------------------
            # Adaptive K by BIC
            # ------------------------------------------------

            best_model = None
            best_bic = np.inf

            for k_candidate in candidate_components:

                try:

                    candidate_model = GaussianMixture(
                        n_components=k_candidate,
                        random_state=(
                            repeat_seed
                            + fold_idx
                            + k_candidate
                        ),
                        reg_covar=1e-6,
                        n_init=5
                    )

                    candidate_model.fit(
                        Xk.reshape(-1, 1)
                    )

                    bic_value = candidate_model.bic(
                        Xk.reshape(-1, 1)
                    )

                    if bic_value < best_bic:
                        best_bic = bic_value
                        best_model = candidate_model

                except Exception:
                    continue

            # ------------------------------------------------
            # Robust fallback for GMM
            # ------------------------------------------------

            if best_model is None:

                fallback_k = min(
                    max(2, int(np.sqrt(target_n))),
                    max_components
                )

                best_model = GaussianMixture(
                    n_components=fallback_k,
                    random_state=(
                        repeat_seed
                        + fold_idx
                    ),
                    reg_covar=1e-6,
                    n_init=5
                )

                best_model.fit(
                    Xk.reshape(-1, 1)
                )

            model = best_model

            # ------------------------------------------------
            # GMM assignment and distance from component mean
            # ------------------------------------------------

            yhat = model.predict(
                Xk.reshape(-1, 1)
            )

            # ------------------------------------------------
            # STANDARDIZED GMM DISTANCE
            #
            # Instead of using only the raw distance from the
            # component mean:
            #
            #     |r_i - mu_k|
            #
            # we normalize it by the standard deviation of the
            # assigned GMM component:
            #
            #     |r_i - mu_k| / (sigma_k + eps)
            #
            # This makes residual deviations comparable across
            # components with different variances.
            #
            # IMPORTANT:
            # This is the ONLY methodological change relative
            # to the attached version. The rest of the method,
            # including the 90/10 residual-density score,
            # second-tier selection, GMM/BIC, and experimental
            # framework is intentionally unchanged.
            # ------------------------------------------------

            component_std = np.sqrt(
                np.maximum(
                    np.asarray(
                        model.covariances_
                    ).reshape(-1),
                    1e-12
                )
            )

            assigned_std = (
                component_std[yhat]
                + 1e-8
            )

            distances = np.abs(
                Xk
                - model.means_[yhat, 0]
            ) / assigned_std

            # ------------------------------------------------
            # Extended majority records
            # ------------------------------------------------

            delete_list3 = []

            for pos in range(len(delete_lists)):

                delete_list3.append([
                    delete_lists[pos][0],
                    delete_lists[pos][1],
                    delete_lists[pos][2],
                    delete_lists[pos][3],
                    delete_lists[pos][4],
                    float(distances[pos]),
                    int(yhat[pos])
                ])

            # ------------------------------------------------
            # Group majority samples by GMM component
            # ------------------------------------------------

            cluster_members = {}

            for cluster_id in np.unique(yhat):

                members = np.where(
                    yhat == cluster_id
                )[0]

                if len(members) > 0:
                    cluster_members[int(cluster_id)] = members

            # ------------------------------------------------
            # If K > target_n, retain components with the
            # strongest residual spread.
            # ------------------------------------------------

            if len(cluster_members) > target_n:

                ordered_clusters = sorted(
                    cluster_members.keys(),
                    key=lambda cid: np.max(
                        distances[cluster_members[cid]]
                    ),
                    reverse=True
                )

                cluster_members = {
                    cid: cluster_members[cid]
                    for cid in ordered_clusters[:target_n]
                }

            active_clusters = list(
                cluster_members.keys()
            )

            # ------------------------------------------------
            # One guaranteed sample per active cluster
            # ------------------------------------------------

            quotas = {
                cid: 1
                for cid in active_clusters
            }

            remaining_quota = (
                target_n
                - len(active_clusters)
            )

            # ------------------------------------------------
            # Proportional quota allocation
            # ------------------------------------------------

            if remaining_quota > 0:

                cluster_sizes = np.asarray([
                    len(cluster_members[cid])
                    for cid in active_clusters
                ], dtype=float)

                proportional = (
                    cluster_sizes
                    / cluster_sizes.sum()
                    * remaining_quota
                )

                base_extra = np.floor(
                    proportional
                ).astype(int)

                for j, cid in enumerate(active_clusters):
                    quotas[cid] += int(base_extra[j])

                leftover = (
                    remaining_quota
                    - int(base_extra.sum())
                )

                order = np.argsort(
                    -(proportional - base_extra)
                )

                for j in order[:leftover]:
                    quotas[active_clusters[j]] += 1

            # ------------------------------------------------
            # TAIL-CAPPED RESIDUAL + DENSITY SELECTION
            #
            # NEW METHOD:
            #
            # The previous version removed the most extreme
            # residual tail before scoring. That can discard
            # genuinely informative hard-majority samples.
            #
            # Here we keep ALL samples in each GMM component,
            # but cap the standardized GMM distance at the
            # component 90th percentile:
            #
            #     r_cap = min(r_i, Q_0.90)
            #
            # Therefore:
            #
            #   - informative hard samples remain eligible;
            #   - isolated extreme outliers cannot dominate;
            #   - residual hardness remains the main signal;
            #   - feature-space density remains the 10% stabilizer.
            #
            # Score_i =
            #     0.90 * CappedResidual_i
            #   + 0.10 * FeatureDensity_i
            #
            # This is a SOFT tail treatment rather than the
            # previous HARD tail exclusion.
            # ------------------------------------------------

            selected_ids = set()

            for cid in active_clusters:

                members = cluster_members[cid]

                cluster_size = len(members)

                quota = min(
                    quotas[cid],
                    cluster_size
                )

                if quota <= 0:
                    continue

                member_distances = np.asarray(
                    [
                        distances[idx]
                        for idx in members
                    ],
                    dtype=float
                )

                # ------------------------------------------------
                # Soft extreme-tail protection
                # ------------------------------------------------

                if len(member_distances) > 1:

                    tail_cap = np.quantile(
                        member_distances,
                        TAIL_CAP_QUANTILE
                    )

                    capped_distances = np.minimum(
                        member_distances,
                        tail_cap
                    )

                else:

                    capped_distances = (
                        member_distances.copy()
                    )

                # ------------------------------------------------
                # Feature-space density for ALL component members
                # ------------------------------------------------

                component_feats = Xk_feat_std[
                    members
                ]

                if len(members) > 1:

                    feat_diff = (
                        component_feats[:, None, :]
                        - component_feats[None, :, :]
                    )

                    feat_distance_matrix = np.sqrt(
                        (feat_diff ** 2).sum(axis=2)
                    )

                    np.fill_diagonal(
                        feat_distance_matrix,
                        np.inf
                    )

                    k_density = min(
                        5,
                        len(members) - 1
                    )

                    nearest_distances = np.partition(
                        feat_distance_matrix,
                        k_density - 1,
                        axis=1
                    )[:, :k_density]

                    mean_knn_distance = np.mean(
                        nearest_distances,
                        axis=1
                    )

                    density_values = 1.0 / (
                        mean_knn_distance
                        + 1e-8
                    )

                else:

                    density_values = np.ones(
                        len(members),
                        dtype=float
                    )

                # ------------------------------------------------
                # Min-max normalization
                # ------------------------------------------------

                residual_min = (
                    capped_distances.min()
                )

                residual_max = (
                    capped_distances.max()
                )

                if (
                    residual_max
                    - residual_min
                    > 1e-12
                ):

                    normalized_residual = (
                        capped_distances
                        - residual_min
                    ) / (
                        residual_max
                        - residual_min
                    )

                else:

                    normalized_residual = np.ones(
                        len(members),
                        dtype=float
                    )

                density_min = (
                    density_values.min()
                )

                density_max = (
                    density_values.max()
                )

                if (
                    density_max
                    - density_min
                    > 1e-12
                ):

                    normalized_density = (
                        density_values
                        - density_min
                    ) / (
                        density_max
                        - density_min
                    )

                else:

                    normalized_density = np.ones(
                        len(members),
                        dtype=float
                    )

                robust_scores = (
                    RESIDUAL_SCORE_WEIGHT
                    * normalized_residual
                    +
                    DENSITY_SCORE_WEIGHT
                    * normalized_density
                )

                scored_candidates = sorted(
                    zip(
                        members.tolist(),
                        robust_scores
                    ),
                    key=lambda item: item[1],
                    reverse=True
                )

                chosen = [
                    idx
                    for idx, score in
                    scored_candidates[
                        :quota
                    ]
                ]

                for idx in chosen:

                    sample = delete_list3[idx]

                    delete_listf.append(
                        sample
                    )

                    selected_ids.add(
                        sample[0]
                    )

            # ------------------------------------------------
            # Final safety: EXACTLY target_n majority samples
            #
            # If a numerical/rounding condition leaves the
            # selection short, fill from remaining majority
            # samples using the same tail-capped residual +
            # feature-density score.
            # ------------------------------------------------

            if len(delete_listf) < target_n:

                remaining = [
                    sample
                    for sample in delete_list3
                    if sample[0] not in selected_ids
                ]

                needed = (
                    target_n
                    - len(delete_listf)
                )

                if len(remaining) > 0:

                    remaining_indices = [
                        delete_list3.index(sample)
                        for sample in remaining
                    ]

                    remaining_distances = np.asarray(
                        [
                            sample[5]
                            for sample in remaining
                        ],
                        dtype=float
                    )

                    if len(remaining_distances) > 1:

                        global_cap = np.quantile(
                            remaining_distances,
                            TAIL_CAP_QUANTILE
                        )

                        capped_remaining = np.minimum(
                            remaining_distances,
                            global_cap
                        )

                    else:

                        capped_remaining = (
                            remaining_distances.copy()
                        )

                    remaining_feats = Xk_feat_std[
                        remaining_indices
                    ]

                    if len(remaining_indices) > 1:

                        feat_diff = (
                            remaining_feats[:, None, :]
                            - remaining_feats[None, :, :]
                        )

                        feat_distance_matrix = np.sqrt(
                            (feat_diff ** 2).sum(axis=2)
                        )

                        np.fill_diagonal(
                            feat_distance_matrix,
                            np.inf
                        )

                        k_density = min(
                            5,
                            len(remaining_indices) - 1
                        )

                        nearest_distances = np.partition(
                            feat_distance_matrix,
                            k_density - 1,
                            axis=1
                        )[:, :k_density]

                        mean_knn_distance = np.mean(
                            nearest_distances,
                            axis=1
                        )

                        remaining_density = 1.0 / (
                            mean_knn_distance
                            + 1e-8
                        )

                    else:

                        remaining_density = np.ones(
                            len(remaining_indices),
                            dtype=float
                        )

                    r_min = capped_remaining.min()
                    r_max = capped_remaining.max()

                    if r_max - r_min > 1e-12:

                        r_norm = (
                            capped_remaining - r_min
                        ) / (
                            r_max - r_min
                        )

                    else:

                        r_norm = np.ones(
                            len(remaining_indices),
                            dtype=float
                        )

                    d_min = remaining_density.min()
                    d_max = remaining_density.max()

                    if d_max - d_min > 1e-12:

                        d_norm = (
                            remaining_density - d_min
                        ) / (
                            d_max - d_min
                        )

                    else:

                        d_norm = np.ones(
                            len(remaining_indices),
                            dtype=float
                        )

                    remaining_scores = (
                        RESIDUAL_SCORE_WEIGHT * r_norm
                        +
                        DENSITY_SCORE_WEIGHT * d_norm
                    )

                    fallback_order = np.argsort(
                        -remaining_scores
                    )

                    take = min(
                        needed,
                        len(fallback_order)
                    )

                    for pos in fallback_order[:take]:

                        sample = remaining[pos]

                        delete_listf.append(
                            sample
                        )

                        selected_ids.add(
                            sample[0]
                        )

            elif len(delete_listf) > target_n:

                # Extremely rare safety condition.
                # Keep the highest-scoring selected samples.
                selected_sorted = sorted(
                    delete_listf,
                    key=lambda sample: sample[5],
                    reverse=True
                )

                delete_listf = (
                    selected_sorted[
                        :target_n
                    ]
                )

        else:
            delete_listf = []

        # ----------------------------------------------------
        # Selected majority
        # ----------------------------------------------------

        delete_listsf2 = sorted(
            delete_listf,
            key=lambda x: x[4]
        )


        # ----------------------------------------------------
        # Selected majority + ALL minority
        # ----------------------------------------------------

        all_list = (
            delete_listsf2
            +
            keep_list
        )


        ret1 = sorted(
            all_list,
            key=lambda x: x[2]
        )


        sampled = ret1


        # ----------------------------------------------------
        # Sampled data
        # ----------------------------------------------------

        sampled_weight = [
            weight[s[0]]
            for s in sampled
        ]


        sampled_X = [
            s[1]
            for s in sampled
        ]


        sampled_Y = [
            s[2]
            for s in sampled
        ]


        sampled_ytt = [
            s[3]
            for s in sampled
        ]


        sampled_err = [
            s[4]
            for s in sampled
        ]


        sampled_X = np.asarray(
            sampled_X,
            dtype=float
        )


        sampled_Y = np.asarray(
            sampled_Y,
            dtype=float
        )


        sampled_ytt = np.asarray(
            sampled_ytt,
            dtype=float
        )


        # ----------------------------------------------------
        # Train weak learner
        # ----------------------------------------------------

        new_predictor.fit(
            sampled_X,
            sampled_ytt
        )


        # ----------------------------------------------------
        # Update error
        # ----------------------------------------------------

        train_prediction = (
            new_predictor.predict(
                Xg
            )
        )


        errc = np.abs(
            ytt
            -
            train_prediction
        )


        # ----------------------------------------------------
        # Update prediction
        # ----------------------------------------------------

        y_pred_traing += (
            train_prediction
            *
            learning_rate
        )


        # ----------------------------------------------------
        # Relative error
        # ----------------------------------------------------

        relative_error = np.abs(
            (
                ytt
                -
                train_prediction
            )
            /
            (
                ytt
                +
                1e-12
            )
        )


        y_error = (
            relative_error
            >
            THRESHOLD
        ).astype(int)


        # ----------------------------------------------------
        # Epsilon
        # ----------------------------------------------------

        epsilon = np.sum(
            y_error
            *
            np.asarray(
                weight
            )
        ) / np.sum(
            weight
        )


        epsilon = np.clip(
            epsilon,
            1e-15,
            1 - 1e-15
        )


        # ----------------------------------------------------
        # Beta
        # ----------------------------------------------------

        beta = (
            epsilon ** 2
        )


        # ----------------------------------------------------
        # Correct samples
        # ----------------------------------------------------

        y_correct = (
            relative_error
            <=
            THRESHOLD
        ).astype(int)


        # ----------------------------------------------------
        # Weight update
        # ----------------------------------------------------

        weight = (
            np.asarray(
                weight
            )
            *
            np.power(
                beta,
                y_correct
            )
        )


        weight = (
            weight
            /
            np.sum(
                weight
            )
        )


        predictors.append(
            new_predictor
        )


        h += 1


    # ========================================================
    # TEST PREDICTION
    # ========================================================

    raw_predictions = np.full(
        (
            len(X_test),
        ),
        ve,
        dtype=float
    )


    for predictor in predictors:

        raw_predictions += (
            predictor.predict(
                X_test
            )
            *
            learning_rate
        )


    # --------------------------------------------------------
    # Probability
    # --------------------------------------------------------

    proba_positive_class = sigmoid(
        raw_predictions
    )


    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    ypp = (
        proba_positive_class
        >
        0.5
    ).astype(int)


    return (
        ypp,
        proba_positive_class
    )


# ============================================================
# 19. METHOD ORDER
# ============================================================

METHOD_DISPLAY_ORDER = [

    "Original GB",

    "Random US",

    "Cluster US",

    "SMOTE",

    "Residual-Guided GMM"
]


# ============================================================
# 20. PRINT FOLD METRICS
# ============================================================

def print_fold_metrics(
    dataset_name,
    repeat_idx,
    fold_idx,
    method_results
):

    print()
    print("=" * 130)

    print(
        f"{dataset_name} | "
        f"Repeat {repeat_idx}/{N_REPEATS} | "
        f"Fold {fold_idx}/{N_FOLDS}"
    )

    print("=" * 130)


    header = (
        f"{'Method':<28}"
        f"{'Acc':>9}"
        f"{'BalAcc':>10}"
        f"{'GMean':>10}"
        f"{'F1':>9}"
        f"{'MCC':>9}"
        f"{'AUC':>9}"
        f"{'Sens':>9}"
        f"{'Spec':>9}"
        f"{'MinRec':>10}"
    )


    print(
        header
    )

    print(
        "-" * 130
    )


    for method in METHOD_DISPLAY_ORDER:

        r = method_results[
            method
        ]


        print(
            f"{method:<28}"
            f"{r['Accuracy']:>9.4f}"
            f"{r['Balanced_Accuracy']:>10.4f}"
            f"{r['GMean']:>10.4f}"
            f"{r['F1']:>9.4f}"
            f"{r['MCC']:>9.4f}"
            f"{r['ROC_AUC']:>9.4f}"
            f"{r['Sensitivity']:>9.4f}"
            f"{r['Specificity']:>9.4f}"
            f"{r['Minority_Recall']:>10.4f}"
        )


# ============================================================
# 21. MAIN EXPERIMENT
# ============================================================

all_fold_results = []

all_dataset_summaries = []

all_repeat_summaries = []

configuration_rows = []


for dataset_name in DATASETS:

    # --------------------------------------------------------
    # Check download
    # --------------------------------------------------------

    if dataset_name not in DATASET_PATHS:

        print()
        print(
            f"[SKIPPED] {dataset_name}"
        )

        continue


    dataset_path = (
        DATASET_PATHS[
            dataset_name
        ]
    )


    # --------------------------------------------------------
    # Load
    # --------------------------------------------------------

    try:

        (
            Xt2,
            Yt2,
            minClass,
            maxClass
        ) = prepare_dataset(
            dataset_path
        )

    except Exception as e:

        print()
        print(
            f"ERROR loading "
            f"{dataset_name}"
        )

        print(e)

        continue


    y = Yt2


    # --------------------------------------------------------
    # Configuration
    # --------------------------------------------------------

    counts = Counter(
        y
    )


    configuration_rows.append({

        "Dataset":
            dataset_name,

        "Samples":
            Xt2.shape[0],

        "Features":
            Xt2.shape[1],

        "Majority_Count":
            counts[0],

        "Minority_Count":
            counts[1],

        "Imbalance_Ratio":
            counts[0] /
            counts[1],

        "Validation":
            "Stratified 5-Fold × 5 Repeats",

        "Repeats":
            N_REPEATS,

        "Folds":
            N_FOLDS,

        "Base_Seed":
            BASE_SEED,

        "N_ITER":
            N_ITER,

        "Actual_Iterations":
            N_ITER + 1,

        "Learning_Rate":
            LEARNING_RATE,

        "Tree_Depth":
            TREE_DEPTH,

        "Threshold":
            THRESHOLD,

        "Proposed_Method":
            (
                "Residual-Guided GMM "
                "Undersampling for "
                "Gradient Boosting"
            ),

        "GMM_Input":
            "Majority residual/error values",

        "Cluster_Selection":
            "Adaptive-BIC GMM + proportional cluster quota + farthest residual samples",

        "Sampling":
            (
                "Selected majority + "
                "all minority"
            )
    })


    # ========================================================
    # REPEATED STRATIFIED CV
    # ========================================================

    for kt2 in range(
        N_REPEATS
    ):

        print()
        print("#" * 110)

        print(
            f"{dataset_name.upper()} | "
            f"REPEAT {kt2 + 1}/{N_REPEATS}"
        )

        print("#" * 110)


        repeat_seed = (
            BASE_SEED
            +
            kt2
        )


        skf = StratifiedKFold(
            n_splits=N_FOLDS,
            shuffle=True,
            random_state=repeat_seed
        )


        # ----------------------------------------------------
        # Temporary storage for this Repeat
        # ----------------------------------------------------

        repeat_results = {
            method: []
            for method
            in METHOD_DISPLAY_ORDER
        }


        # ====================================================
        # FOLD LOOP
        # ====================================================

        for it2, (
            train_index,
            test_index
        ) in enumerate(
            skf.split(
                Xt2,
                Yt2
            ),
            start=1
        ):


            print()
            print(
                f"Processing "
                f"{dataset_name} | "
                f"Repeat {kt2 + 1}/{N_REPEATS} | "
                f"Fold {it2}/{N_FOLDS}"
            )


            # ------------------------------------------------
            # Train / Test
            # ------------------------------------------------

            X_train = Xt2[
                train_index
            ]

            X_test = Xt2[
                test_index
            ]

            y_train = Yt2[
                train_index
            ]

            y_test = Yt2[
                test_index
            ]


            fold_seed = (
                repeat_seed
                +
                it2
            )


            method_results = {}


            # =================================================
            # 1. ORIGINAL GB
            # =================================================

            pred_original, prob_original = (
                train_standard_gb(
                    X_train,
                    y_train,
                    X_test,
                    fold_seed
                )
            )


            original_results = calculate_metrics(
                y_test,
                pred_original,
                prob_original,
                minClass
            )


            method_results[
                "Original GB"
            ] = original_results


            # =================================================
            # 2. RANDOM US
            # =================================================

            X_random, y_random = (
                random_undersample(
                    X_train,
                    y_train,
                    fold_seed
                )
            )


            pred_random, prob_random = (
                train_standard_gb(
                    X_random,
                    y_random,
                    X_test,
                    fold_seed
                )
            )


            random_results = calculate_metrics(
                y_test,
                pred_random,
                prob_random,
                minClass
            )


            method_results[
                "Random US"
            ] = random_results


            # =================================================
            # 3. CLUSTER US
            # =================================================

            X_cluster, y_cluster = (
                cluster_undersample(
                    X_train,
                    y_train,
                    fold_seed
                )
            )


            pred_cluster, prob_cluster = (
                train_standard_gb(
                    X_cluster,
                    y_cluster,
                    X_test,
                    fold_seed
                )
            )


            cluster_results = calculate_metrics(
                y_test,
                pred_cluster,
                prob_cluster,
                minClass
            )


            method_results[
                "Cluster US"
            ] = cluster_results


            # =================================================
            # 4. SMOTE
            # =================================================

            X_smote, y_smote = (
                smote_resample(
                    X_train,
                    y_train,
                    fold_seed
                )
            )


            pred_smote, prob_smote = (
                train_standard_gb(
                    X_smote,
                    y_smote,
                    X_test,
                    fold_seed
                )
            )


            smote_results = calculate_metrics(
                y_test,
                pred_smote,
                prob_smote,
                minClass
            )


            method_results[
                "SMOTE"
            ] = smote_results


            # =================================================
            # 5. PROPOSED
            # =================================================

            pred_proposed, prob_proposed = (
                train_proposed_cell60(
                    X_train,
                    y_train,
                    X_test,
                    repeat_seed,
                    it2
                )
            )


            proposed_results = calculate_metrics(
                y_test,
                pred_proposed,
                prob_proposed,
                minClass
            )


            method_results[
                "Residual-Guided GMM"
            ] = proposed_results


            # =================================================
            # PRINT ALL METRICS
            # =================================================

            print_fold_metrics(
                dataset_name,
                kt2 + 1,
                it2,
                method_results
            )


            # =================================================
            # STORE FOLD RESULTS
            # =================================================

            for method_name in METHOD_DISPLAY_ORDER:

                result = (
                    method_results[
                        method_name
                    ]
                )


                # --------------------------------------------
                # Save for repeat averaging
                # --------------------------------------------

                repeat_results[
                    method_name
                ].append(
                    result
                )


                # --------------------------------------------
                # Save fold result
                # --------------------------------------------

                row = {

                    "Dataset":
                        dataset_name,

                    "Repeat":
                        kt2 + 1,

                    "Fold":
                        it2,

                    "Method":
                        method_name,

                    "Accuracy":
                        result[
                            "Accuracy"
                        ],

                    "Balanced_Accuracy":
                        result[
                            "Balanced_Accuracy"
                        ],

                    "GMean":
                        result[
                            "GMean"
                        ],

                    "F1":
                        result[
                            "F1"
                        ],

                    "MCC":
                        result[
                            "MCC"
                        ],

                    "ROC_AUC":
                        result[
                            "ROC_AUC"
                        ],

                    "Sensitivity":
                        result[
                            "Sensitivity"
                        ],

                    "Specificity":
                        result[
                            "Specificity"
                        ],

                    "Minority_Recall":
                        result[
                            "Minority_Recall"
                        ],

                    "TN":
                        result[
                            "TN"
                        ],

                    "FP":
                        result[
                            "FP"
                        ],

                    "FN":
                        result[
                            "FN"
                        ],

                    "TP":
                        result[
                            "TP"
                        ]
                }


                all_fold_results.append(
                    row
                )


        # ====================================================
        # END OF REPEAT
        #
        # IMPORTANT:
        #
        # First calculate mean across 5 folds.
        # ====================================================

        repeat_summary_metrics = [

            "Accuracy",

            "Balanced_Accuracy",

            "GMean",

            "F1",

            "MCC",

            "ROC_AUC",

            "Sensitivity",

            "Specificity",

            "Minority_Recall"
        ]


        for method in METHOD_DISPLAY_ORDER:

            repeat_row = {

                "Dataset":
                    dataset_name,

                "Repeat":
                    kt2 + 1,

                "Method":
                    method
            }


            for metric in (
                repeat_summary_metrics
            ):

                values = [

                    x[metric]
                    for x
                    in
                    repeat_results[
                        method
                    ]
                ]


                repeat_mean = np.nanmean(
                    values
                )


                repeat_row[
                    metric
                ] = repeat_mean


            all_repeat_summaries.append(
                repeat_row
            )


        # ====================================================
        # PRINT REPEAT SUMMARY
        # ====================================================

        repeat_df_temp = pd.DataFrame(
            [
                r
                for r
                in all_repeat_summaries
                if (
                    r["Dataset"]
                    ==
                    dataset_name
                    and
                    r["Repeat"]
                    ==
                    kt2 + 1
                )
            ]
        )


        print()
        print("=" * 130)

        print(
            f"{dataset_name} | "
            f"REPEAT {kt2 + 1} SUMMARY "
            f"(Mean over {N_FOLDS} folds)"
        )

        print("=" * 130)


        display_cols = [

            "Method",

            "Accuracy",

            "Balanced_Accuracy",

            "GMean",

            "F1",

            "MCC",

            "ROC_AUC",

            "Sensitivity",

            "Specificity",

            "Minority_Recall"
        ]


        print(
            repeat_df_temp[
                display_cols
            ].to_string(
                index=False,
                float_format=lambda x:
                f"{x:.4f}"
            )
        )


    # ========================================================
    # DATASET-LEVEL SUMMARY
    #
    # IMPORTANT:
    #
    # Mean over folds
    #       ↓
    # Repeat mean
    #       ↓
    # Mean ± SD over repeats
    #
    # This is NOT a direct pooling of 25 folds.
    # ========================================================

    repeat_df_dataset = pd.DataFrame(
        [
            r
            for r
            in all_repeat_summaries
            if r[
                "Dataset"
            ]
            ==
            dataset_name
        ]
    )


    summary_metrics = [

        "Accuracy",

        "Balanced_Accuracy",

        "GMean",

        "F1",

        "MCC",

        "ROC_AUC",

        "Sensitivity",

        "Specificity",

        "Minority_Recall"
    ]


    dataset_summary_rows = []


    for method in METHOD_DISPLAY_ORDER:

        method_repeat_df = (
            repeat_df_dataset[
                repeat_df_dataset[
                    "Method"
                ]
                ==
                method
            ]
        )


        row = {

            "Dataset":
                dataset_name,

            "Method":
                method
        }


        for metric in summary_metrics:

            mean_value = (
                method_repeat_df[
                    metric
                ].mean()
            )


            std_value = (
                method_repeat_df[
                    metric
                ].std(
                    ddof=1
                )
            )


            row[
                metric
            ] = (
                f"{mean_value:.4f} ± "
                f"{std_value:.4f}"
            )


            row[
                metric + "_Mean"
            ] = mean_value


            row[
                metric + "_Std"
            ] = std_value


        dataset_summary_rows.append(
            row
        )


    all_dataset_summaries.extend(
        dataset_summary_rows
    )


    dataset_summary_df = pd.DataFrame(
        dataset_summary_rows
    )


    # ========================================================
    # PRINT DATASET-LEVEL RESULT
    # ========================================================

    print()
    print()
    print("=" * 140)

    print(
        f"DATASET-LEVEL SUMMARY: {dataset_name}"
    )

    print(
        "Mean over folds → mean over repeats → Mean ± SD"
    )

    print("=" * 140)


    print(
        dataset_summary_df[
            [
                "Method"
            ]
            +
            summary_metrics
        ].to_string(
            index=False
        )
    )


# ============================================================
# 22. CREATE DATAFRAMES
# ============================================================

fold_results_df = pd.DataFrame(
    all_fold_results
)


repeat_results_df = pd.DataFrame(
    all_repeat_summaries
)


summary_df = pd.DataFrame(
    all_dataset_summaries
)


configuration_df = pd.DataFrame(
    configuration_rows
)


# ============================================================
# 23. CHECK RESULTS
# ============================================================

if len(
    fold_results_df
) == 0:

    raise RuntimeError(
        "No experiment results were generated."
    )


# ============================================================
# 24. SAVE FOLD RESULTS
# ============================================================

fold_results_df.to_csv(
    "fold_results.csv",
    index=False
)


print()
print(
    "Saved: fold_results.csv"
)


# ============================================================
# 25. SAVE REPEAT RESULTS
# ============================================================

repeat_results_df.to_csv(
    "repeat_results.csv",
    index=False
)


print(
    "Saved: repeat_results.csv"
)


# ============================================================
# 26. SAVE DATASET-LEVEL SUMMARY
# ============================================================

summary_df.to_csv(
    "summary_mean_std.csv",
    index=False
)


print(
    "Saved: summary_mean_std.csv"
)


# ============================================================
# 27. METHOD COMPARISON
# ============================================================

comparison_columns = [

    "Dataset",

    "Method",

    "Accuracy",

    "Balanced_Accuracy",

    "GMean",

    "F1",

    "MCC",

    "ROC_AUC",

    "Sensitivity",

    "Specificity",

    "Minority_Recall"
]


method_comparison_df = (
    summary_df[
        comparison_columns
    ].copy()
)


method_comparison_df.to_csv(
    "method_comparison.csv",
    index=False
)


print(
    "Saved: method_comparison.csv"
)


# ============================================================
# 28. CONFUSION MATRICES
# ============================================================

confusion_rows = []


for dataset_name in DATASETS:

    for method in METHOD_DISPLAY_ORDER:

        subset = fold_results_df[
            (
                fold_results_df[
                    "Dataset"
                ]
                ==
                dataset_name
            )
            &
            (
                fold_results_df[
                    "Method"
                ]
                ==
                method
            )
        ]


        if len(
            subset
        ) == 0:

            continue


        confusion_rows.append({

            "Dataset":
                dataset_name,

            "Method":
                method,

            "TN":
                int(
                    subset[
                        "TN"
                    ].sum()
                ),

            "FP":
                int(
                    subset[
                        "FP"
                    ].sum()
                ),

            "FN":
                int(
                    subset[
                        "FN"
                    ].sum()
                ),

            "TP":
                int(
                    subset[
                        "TP"
                    ].sum()
                )
        })


confusion_df = pd.DataFrame(
    confusion_rows
)


confusion_df.to_csv(
    "confusion_matrices.csv",
    index=False
)


print(
    "Saved: confusion_matrices.csv"
)


# ============================================================
# 29. DATASET CONFIGURATION
# ============================================================

configuration_df.to_csv(
    "dataset_configuration.csv",
    index=False
)


print(
    "Saved: dataset_configuration.csv"
)


# ============================================================
# 30. DATASET-LEVEL RANKING
#
# Main imbalance-aware metrics:
#
# Balanced Accuracy
# G-Mean
# F1
# MCC
# ROC-AUC
# Minority Recall
#
# Higher = better
# ============================================================

ranking_metrics = [

    "Balanced_Accuracy_Mean",

    "GMean_Mean",

    "F1_Mean",

    "MCC_Mean",

    "ROC_AUC_Mean",

    "Minority_Recall_Mean"
]


ranking_rows = []


for dataset_name in DATASETS:

    dataset_summary = (
        summary_df[
            summary_df[
                "Dataset"
            ]
            ==
            dataset_name
        ].copy()
    )


    if len(
        dataset_summary
    ) == 0:

        continue


    for metric in ranking_metrics:

        dataset_summary[
            metric + "_Rank"
        ] = (
            dataset_summary[
                metric
            ]
            .rank(
                ascending=False,
                method="average"
            )
        )


    rank_columns = [

        metric + "_Rank"
        for metric
        in ranking_metrics
    ]


    dataset_summary[
        "Average_Rank"
    ] = (
        dataset_summary[
            rank_columns
        ]
        .mean(
            axis=1
        )
    )


    ranking_rows.append(
        dataset_summary[
            [
                "Dataset",
                "Method",
                "Average_Rank"
            ]
        ]
    )


if ranking_rows:

    ranking_df = pd.concat(
        ranking_rows,
        ignore_index=True
    )

else:

    ranking_df = pd.DataFrame(
        columns=[
            "Dataset",
            "Method",
            "Average_Rank"
        ]
    )


ranking_df.to_csv(
    "method_ranking_by_dataset.csv",
    index=False
)


print(
    "Saved: method_ranking_by_dataset.csv"
)


# ============================================================
# 31. OVERALL ACROSS DATASETS
#
# VERY IMPORTANT:
#
# We first have:
#
# Fold means
#      ↓
# Repeat means
#      ↓
# Dataset mean
#
# NOW:
#
# Dataset means
#      ↓
# Overall mean across datasets
#
# Therefore datasets have equal weight.
#
# A dataset with 10,000 samples does NOT dominate
# a dataset with 500 samples.
# ============================================================

overall_rows = []


for method in METHOD_DISPLAY_ORDER:

    method_dataset_df = (
        summary_df[
            summary_df[
                "Method"
            ]
            ==
            method
        ]
    )


    row = {

        "Method":
            method
    }


    for metric in summary_metrics:

        mean_column = (
            metric
            +
            "_Mean"
        )


        values = (
            method_dataset_df[
                mean_column
            ]
            .dropna()
        )


        if len(values) > 0:

            overall_mean = (
                values.mean()
            )

            overall_std = (
                values.std(
                    ddof=1
                )
                if len(values) > 1
                else 0.0
            )

        else:

            overall_mean = np.nan

            overall_std = np.nan


        row[
            metric
        ] = (
            f"{overall_mean:.4f} ± "
            f"{overall_std:.4f}"
        )


        row[
            metric + "_Mean"
        ] = overall_mean


        row[
            metric + "_SD"
        ] = overall_std


        row[
            metric + "_N_Datasets"
        ] = len(values)


    overall_rows.append(
        row
    )


overall_df = pd.DataFrame(
    overall_rows
)


# ============================================================
# 32. PRINT OVERALL RESULTS
# ============================================================

print()
print()
print("=" * 150)

print(
    "OVERALL RESULTS ACROSS DATASETS"
)

print(
    f"Number of datasets actually evaluated: "
    f"{len(summary_df['Dataset'].unique())}"
)

print(
    "Each dataset has equal weight."
)

print(
    "Dataset mean → Overall mean"
)

print("=" * 150)


overall_display_columns = [

    "Method",

    "Accuracy",

    "Balanced_Accuracy",

    "GMean",

    "F1",

    "MCC",

    "ROC_AUC",

    "Sensitivity",

    "Specificity",

    "Minority_Recall"
]


print(
    overall_df[
        overall_display_columns
    ].to_string(
        index=False
    )
)


# ============================================================
# 33. SAVE OVERALL RESULTS
# ============================================================

overall_df.to_csv(
    "overall_across_datasets.csv",
    index=False
)


print()
print(
    "Saved: overall_across_datasets.csv"
)


# ============================================================
# 34. PROPOSED WINS
#
# For every dataset:
#
# Compare the DATASET-LEVEL MEAN of the methods.
#
# Proposed wins if:
#
# Proposed > every other method
#
# for the given metric.
#
# Ties are NOT counted as wins.
# ============================================================

WIN_METRICS = {

    "Balanced Accuracy":
        "Balanced_Accuracy_Mean",

    "G-Mean":
        "GMean_Mean",

    "F1":
        "F1_Mean",

    "MCC":
        "MCC_Mean",

    "AUC":
        "ROC_AUC_Mean"
}


proposed_win_rows = []


for metric_name, metric_column in (
    WIN_METRICS.items()
):


    wins = 0

    evaluated_datasets = 0


    for dataset_name in (
        summary_df[
            "Dataset"
        ].unique()
    ):


        dataset_data = (
            summary_df[
                summary_df[
                    "Dataset"
                ]
                ==
                dataset_name
            ]
        )


        proposed_row = (
            dataset_data[
                dataset_data[
                    "Method"
                ]
                ==
                "Residual-Guided GMM"
            ]
        )


        if len(
            proposed_row
        ) == 0:

            continue


        proposed_value = (
            proposed_row.iloc[0][
                metric_column
            ]
        )


        other_values = (
            dataset_data[
                dataset_data[
                    "Method"
                ]
                !=
                "Residual-Guided GMM"
            ][
                metric_column
            ]
            .dropna()
        )


        if len(
            other_values
        ) == 0:

            continue


        evaluated_datasets += 1


        # ----------------------------------------------------
        # Strict win
        # ----------------------------------------------------

        if (
            proposed_value
            >
            other_values.max()
        ):

            wins += 1


    proposed_win_rows.append({

        "Metric":
            metric_name,

        "Proposed_Wins":
            wins,

        "Datasets_Evaluated":
            evaluated_datasets,

        "Win_Rate":
            (
                wins /
                evaluated_datasets
                if evaluated_datasets > 0
                else np.nan
            )
    })


proposed_wins_df = pd.DataFrame(
    proposed_win_rows
)


# ============================================================
# 35. PRINT PROPOSED WINS
# ============================================================

print()
print()
print("=" * 100)

print(
    "PROPOSED METHOD WINS"
)

print("=" * 100)


for _, row in (
    proposed_wins_df.iterrows()
):

    print(
        f"{row['Metric']:<22}: "
        f"{int(row['Proposed_Wins'])} / "
        f"{int(row['Datasets_Evaluated'])}"
    )


# ============================================================
# 36. SAVE PROPOSED WINS
# ============================================================

proposed_wins_df.to_csv(
    "proposed_wins.csv",
    index=False
)


print()
print(
    "Saved: proposed_wins.csv"
)


# ============================================================
# 37. BEST METHOD PER DATASET FOR EACH METRIC
# ============================================================

metric_winner_rows = []


for dataset_name in (
    summary_df[
        "Dataset"
    ].unique()
):


    dataset_data = (
        summary_df[
            summary_df[
                "Dataset"
            ]
            ==
            dataset_name
        ]
    )


    for metric_name, metric_column in (
        WIN_METRICS.items()
    ):


        if len(
            dataset_data
        ) == 0:

            continue


        idx = (
            dataset_data[
                metric_column
            ]
            .idxmax()
        )


        winner_method = (
            summary_df.loc[
                idx,
                "Method"
            ]
        )


        winner_value = (
            summary_df.loc[
                idx,
                metric_column
            ]
        )


        metric_winner_rows.append({

            "Dataset":
                dataset_name,

            "Metric":
                metric_name,

            "Winner":
                winner_method,

            "Value":
                winner_value
        })


metric_winners_df = pd.DataFrame(
    metric_winner_rows
)


metric_winners_df.to_csv(
    "metric_winners_by_dataset.csv",
    index=False
)


print(
    "Saved: metric_winners_by_dataset.csv"
)


# ============================================================
# 38. COUNT WINS FOR ALL METHODS
#
# This is useful for the paper because it shows whether
# Proposed wins are actually competitive against every
# baseline.
# ============================================================

all_method_win_rows = []


for metric_name, metric_column in (
    WIN_METRICS.items()
):


    for dataset_name in (
        summary_df[
            "Dataset"
        ].unique()
    ):


        dataset_data = (
            summary_df[
                summary_df[
                    "Dataset"
                ]
                ==
                dataset_name
            ]
        )


        if len(
            dataset_data
        ) == 0:

            continue


        max_value = (
            dataset_data[
                metric_column
            ]
            .max()
        )


        winners = dataset_data[
            np.isclose(
                dataset_data[
                    metric_column
                ],
                max_value,
                rtol=1e-12,
                atol=1e-12
            )
        ]


        # ----------------------------------------------------
        # Ties are recorded separately.
        # ----------------------------------------------------

        if len(
            winners
        ) == 1:

            all_method_win_rows.append({

                "Metric":
                    metric_name,

                "Dataset":
                    dataset_name,

                "Winner":
                    winners.iloc[0][
                        "Method"
                    ],

                "Type":
                    "Unique winner"

            })

        else:

            all_method_win_rows.append({

                "Metric":
                    metric_name,

                "Dataset":
                    dataset_name,

                "Winner":
                    " / ".join(
                        winners[
                            "Method"
                        ].tolist()
                    ),

                "Type":
                    "Tie"

            })


all_metric_winners_df = pd.DataFrame(
    all_method_win_rows
)


all_metric_winners_df.to_csv(
    "all_metric_winners.csv",
    index=False
)


print(
    "Saved: all_metric_winners.csv"
)


# ============================================================
# 39. OVERALL METHOD RANKING
#
# Based on six imbalance-aware metrics:
#
# Balanced Accuracy
# G-Mean
# F1
# MCC
# ROC-AUC
# Minority Recall
# ============================================================

overall_ranking = (
    ranking_df
    .groupby(
        "Method"
    )[
        "Average_Rank"
    ]
    .mean()
    .sort_values()
    .reset_index()
)


overall_ranking.rename(
    columns={
        "Average_Rank":
            "Overall_Average_Rank"
    },
    inplace=True
)


overall_ranking.to_csv(
    "overall_method_ranking.csv",
    index=False
)


# ============================================================
# 40. PRINT OVERALL RANKING
# ============================================================

print()
print()
print("=" * 100)

print(
    "OVERALL METHOD RANKING"
)

print("=" * 100)


print(
    overall_ranking.to_string(
        index=False
    )
)


# ============================================================
# 41. BEST METHOD OVERALL
# ============================================================

best_method = (
    overall_ranking.iloc[0][
        "Method"
    ]
)


best_rank = (
    overall_ranking.iloc[0][
        "Overall_Average_Rank"
    ]
)


print()
print("=" * 100)

print(
    "FINAL OVERALL CONCLUSION"
)

print("=" * 100)


print(
    f"Best overall method by average rank: "
    f"{best_method}"
)


print(
    f"Overall average rank: "
    f"{best_rank:.4f}"
)


# ============================================================
# 42. BEST METHOD FOR EACH DATASET
#
# Based on average rank across:
#
# Balanced Accuracy
# G-Mean
# F1
# MCC
# ROC-AUC
# Minority Recall
# ============================================================

winner_rows = []


for dataset_name in (
    ranking_df[
        "Dataset"
    ].unique()
):


    dataset_rank = (
        ranking_df[
            ranking_df[
                "Dataset"
            ]
            ==
            dataset_name
        ]
    )


    if len(
        dataset_rank
    ) == 0:

        continue


    winner_idx = (
        dataset_rank[
            "Average_Rank"
        ].idxmin()
    )


    winner = (
        ranking_df.loc[
            winner_idx
        ]
    )


    winner_rows.append({

        "Dataset":
            dataset_name,

        "Best_Method":
            winner[
                "Method"
            ],

        "Average_Rank":
            winner[
                "Average_Rank"
            ]

    })


winner_df = pd.DataFrame(
    winner_rows
)


winner_df.to_csv(
    "dataset_winners.csv",
    index=False
)


# ============================================================
# 43. PRINT DATASET WINNERS
# ============================================================

print()
print("=" * 100)

print(
    "BEST METHOD FOR EACH DATASET"
)

print("=" * 100)


print(
    winner_df.to_string(
        index=False
    )
)


# ============================================================
# 44. COUNT DATASET-LEVEL WINS
# ============================================================

win_counts = (
    winner_df[
        "Best_Method"
    ]
    .value_counts()
    .reset_index()
)


win_counts.columns = [

    "Method",

    "Dataset_Wins"
]


print()
print("=" * 100)

print(
    "NUMBER OF DATASET-LEVEL WINS"
)

print("=" * 100)


print(
    win_counts.to_string(
        index=False
    )
)


# ============================================================
# 45. FINAL EXPERIMENT INFORMATION
# ============================================================

actual_datasets = (
    summary_df[
        "Dataset"
    ].nunique()
)


print()
print()
print("=" * 120)

print(
    "EXPERIMENT SUMMARY"
)

print("=" * 120)


print(
    f"Total datasets available : "
    f"{len(ALL_DATASETS)}"
)

print(
    f"Datasets requested       : "
    f"{N_DATASETS_TO_RUN}"
)

print(
    f"Datasets successfully run: "
    f"{actual_datasets}"
)

print(
    f"Methods per dataset      : "
    f"{len(METHOD_DISPLAY_ORDER)}"
)

print(
    f"Folds per repeat         : "
    f"{N_FOLDS}"
)

print(
    f"Repeats per dataset      : "
    f"{N_REPEATS}"
)

print(
    f"Total folds per dataset  : "
    f"{N_FOLDS * N_REPEATS}"
)

print(
    f"Total method-fold runs   : "
    f"{actual_datasets * len(METHOD_DISPLAY_ORDER) * N_FOLDS * N_REPEATS}"
)


# ============================================================
# 46. FINAL OUTPUT FILES
# ============================================================

print()
print()
print("=" * 120)

print(
    "FINAL OUTPUT FILES"
)

print("=" * 120)


output_files = [

    "fold_results.csv",

    "repeat_results.csv",

    "summary_mean_std.csv",

    "method_comparison.csv",

    "confusion_matrices.csv",

    "dataset_configuration.csv",

    "method_ranking_by_dataset.csv",

    "overall_method_ranking.csv",

    "overall_across_datasets.csv",

    "proposed_wins.csv",

    "metric_winners_by_dataset.csv",

    "all_metric_winners.csv",

    "dataset_winners.csv"
]


for i, filename in enumerate(
    output_files,
    start=1
):

    print(
        f"{i:2d}. {filename}"
    )


# ============================================================
# 47. FINAL PAPER-STYLE REPORT
# ============================================================

print()
print()
print("=" * 150)

print(
    "PAPER-STYLE FINAL REPORT"
)

print("=" * 150)


print()
print("LEVEL 1 — DATASET-LEVEL RESULTS")
print("-" * 150)


for dataset_name in (
    summary_df[
        "Dataset"
    ].unique()
):

    print()
    print(
        f"Dataset: {dataset_name}"
    )

    print(
        "-" * 150
    )


    temp = summary_df[
        summary_df[
            "Dataset"
        ]
        ==
        dataset_name
    ]


    print(
        temp[
            [
                "Method",

                "Accuracy",

                "Balanced_Accuracy",

                "GMean",

                "F1",

                "MCC",

                "ROC_AUC",

                "Sensitivity",

                "Specificity",

                "Minority_Recall"
            ]
        ].to_string(
            index=False
        )
    )


print()
print()
print(
    "LEVEL 2 — OVERALL ACROSS DATASETS"
)

print("-" * 150)


print(
    overall_df[
        [
            "Method",

            "Accuracy",

            "Balanced_Accuracy",

            "GMean",

            "F1",

            "MCC",

            "ROC_AUC",

            "Sensitivity",

            "Specificity",

            "Minority_Recall"
        ]
    ].to_string(
        index=False
    )
)


# ============================================================
# 48. PROPOSED WINS — PAPER FORMAT
# ============================================================

print()
print()
print(
    "PROPOSED WINS"
)

print("-" * 80)


for _, row in (
    proposed_wins_df.iterrows()
):

    print(
        f"{row['Metric']:<25}"
        f": "
        f"{int(row['Proposed_Wins'])} / "
        f"{int(row['Datasets_Evaluated'])}"
    )


# ============================================================
# 49. FINAL
# ============================================================

print()
print()
print("=" * 120)

print(
    "EXPERIMENT COMPLETED SUCCESSFULLY"
)

print("=" * 120)

print()
print(
    f"Datasets evaluated: "
    f"{actual_datasets}"
)

print(
    f"Methods: "
    f"{len(METHOD_DISPLAY_ORDER)}"
)

print(
    f"Validation: "
    f"Stratified {N_FOLDS}-Fold × "
    f"{N_REPEATS} Repeats"
)

print()
print(
    "Dataset-level aggregation:"
)

print(
    "    Fold means"
)

print(
    "        ↓"
)

print(
    "    Repeat means"
)

print(
    "        ↓"
)

print(
    "    Dataset Mean ± SD"
)

print()
print(
    "Overall aggregation:"
)

print(
    "    Dataset means"
)

print(
    "        ↓"
)

print(
    "    Overall mean across datasets"
)

print()
print(
    "Proposed win counts are calculated separately "
    "for Balanced Accuracy, G-Mean, F1, MCC and AUC."
)

print()
print(
    "Experiment completed."
)
```
